## Análise de tendência:
O objetivo da análise é avaliar qual a tendência de compra entre determinados grupos de acordo com as suas características e outros fatores. Ao final queremos obter informações valiosas que possa ajudar a loja a realizar vendas mais acertivas.

##### Análise de frequências: Com que frequência a maioria das pessoas compra? & Qual a idade das pessoas que compram com maior frequência?

```python
frequencia = df1['Frequency of Purchases'].value_counts()
frequencia
```

```python
frequencia.plot(kind = 'bar')
```

```python
df1['Frequency of Purchases'].value_counts(normalize=True) * 100
```


> A frquência de compra mais comum é trimestral, tendo 15% (584) do nosso dataset com essa frequência. E apesar da frequência trimestral ser a mais comum, as diferenças entre as categorias são pequenas, indicando uma distribuição relativamente equilibrada entre as frequências de compra
>> Ter essa análise ajuda a loja a se preparar para vários tipos de clientes.

*Para descobrir a frequência de compras de acordo com a idade, consideraremos a frequência semanal que é a maior frequência no dataset.*

```python
frequencia_semanal = df1[df1['Frequency of Purchases'] == 'Weekly']
frequencia_semanal.describe(include='all')
```

```python
frequencia_semanal
```

```python
agrup = frequencia_semanal.groupby(['Gender','Frequency of Purchases'])['Age'].agg(lambda x: x.mode()[0])
#agrup = agrup.sort_values(ascending=False)
agrup
```

```python
agrup.mean()
```

```python
agrup.plot(kind = 'bar')
```

>"Entre os clientes que realizam compras semanalmente, a idade média por gênero é de 38 anos, as idades mais frequentes são 35 para mulheres e 44 para homens."
>> Identificar das faixas etárias mais frequentes entre os compradores semanais pode ajudar a loja a explorar estratégias de fidelização e benefícios direcionados a esse perfil, avaliando se essas ações aumentam a frequência de compra.
